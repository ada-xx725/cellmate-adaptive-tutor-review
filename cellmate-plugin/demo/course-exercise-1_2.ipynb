## Exercise 1.2: Convert from metres to British length units

Convert a length in metres into inches, feet, yards, and miles while keeping the course variable names.

```python
from pathlib import Path
import os
import sys

for root in [Path.cwd(), *Path.cwd().parents]:
    lecture_dir = root / "external" / "introduction-to-python" / "lecture1"
    if (lecture_dir / "lecture.py").exists():
        sys.path.insert(0, str(lecture_dir))
        os.chdir(lecture_dir)
        break
else:
    raise FileNotFoundError("Could not locate the pinned Introduction to Python course checkout.")

import pybryt
from lecture import pybryt_reference
```

```python
metres = 640

# 1 inch = 2.54 cm = 0.0254 m.
inches = metres / 0.0254
feet = inches / 12
yards = feet / 3
miles = yards / 1760
```

```python
with pybryt.check(pybryt_reference(1, 2)):
    feet, yards, miles
```

```python
import numbers
import numpy as np

assert all(isinstance(value, numbers.Real) for value in [feet, yards, miles])
assert np.isclose(feet, 2099.737532808399)
assert np.isclose(yards, 699.912510936133)
assert np.isclose(miles, 0.3976775630318938)
```
